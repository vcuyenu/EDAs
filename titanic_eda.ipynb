Importing necessary libraries (numpy, pandas, seaborn) and using dataframes to show basic information about the dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('titanic.csv')
df.shape
```

Using seaborn to graph boxplots to determine outliers in the set (for age and fare)

```python
sns.boxplot(x = df['Age'])
plt.show()

sns.boxplot(x = df['Fare'])
plt.show()
```

Graphing the relationship between age, those who survived, their class and the price of their tickets

```python
df['Age'].hist()
plt.title("Age Distribution")
plt.show()

sns.countplot(x = 'Survived', data = df)
```

```python
sns.countplot(x= 'Pclass', hue = 'Survived', data = df)
```

```python
sns.countplot(x = 'Sex', hue = 'Survived', data = df)
```

```python
sns.scatterplot(x = 'Age', y = 'Fare', hue = 'Survived', data = df)
```

Replacing  missing values with appropriate measures of tendency (mean for age, mode for those embarked) and checking if any is left null

```python
df['Age'].fillna(df['Age'].mean())
df['Embarked'].fillna(df['Embarked'].mode()[0])
df.isnull().sum()
```
